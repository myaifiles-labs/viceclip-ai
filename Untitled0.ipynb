````python
%%writefile app.py
import os, json, re, uuid, wave, threading, subprocess, traceback, requests
import numpy as np
from fastapi import FastAPI, UploadFile, File, Form
from fastapi.responses import HTMLResponse, JSONResponse
from fastapi.staticfiles import StaticFiles
from faster_whisper import WhisperModel
from google import genai

os.environ["PATH"] += ":/root/.deno/bin"
app = FastAPI(title="ViceClip AI")
for d in ["sessions", "fonts", "static"]:
    os.makedirs(d, exist_ok=True)
app.mount("/sessions", StaticFiles(directory="sessions"), name="sessions")

whisper_model = WhisperModel("base", device="auto", compute_type="int8")
JOBS = {}

FONTS = {
    "Montserrat": "https://github.com/google/fonts/raw/main/ofl/montserrat/Montserrat%5Bwght%5D.ttf",
    "Anton": "https://github.com/google/fonts/raw/main/ofl/anton/Anton-Regular.ttf",
    "Baloo 2": "https://github.com/google/fonts/raw/main/ofl/baloo2/Baloo2%5Bwght%5D.ttf",
    "Poppins": "https://github.com/google/fonts/raw/main/ofl/poppins/Poppins-Bold.ttf",
    "Oswald": "https://github.com/google/fonts/raw/main/ofl/oswald/Oswald%5Bwght%5D.ttf",
    "Bebas Neue": "https://github.com/google/fonts/raw/main/ofl/bebasneue/BebasNeue-Regular.ttf",
}

def ensure_font(name):
    p = f"fonts/{name.replace(' ', '_')}.ttf"
    if not os.path.exists(p):
        r = requests.get(FONTS.get(name, FONTS["Montserrat"]), timeout=15)
        open(p, "wb").write(r.content)
    return p

def hex_ass(h, a="00"):
    h = h.lstrip("#") if len(h.lstrip("#")) == 6 else "FFFFFF"
    return f"&H{a}{h[4:6]}{h[2:4]}{h[0:2]}".upper()

def fmt_t(s):
    s = max(0.0, s)
    h, m, sec, cs = int(s // 3600), int((s % 3600) // 60), int(s % 60), int(round((s - int(s)) * 100))
    if cs >= 100:
        sec += 1
        cs = 0
    return f"{h}:{m:02d}:{sec:02d}.{cs:02d}"

def build_ass(words, st_c, en_c, mode, font, fsize, y_pct, tc, hc, bc, path):
    cp, ch, cb = hex_ass(tc), hex_ass(hc), hex_ass(bc, "25")
    is_karaoke = (mode == "Karaoke Pill Box")
    bs = "3" if is_karaoke else "1"
    osize = "8" if is_karaoke else ("5" if mode == "Neon Vice Glow" else "3")
    sh = "3" if mode == "Neon Vice Glow" else ("0" if is_karaoke else "1")
    back = cb if is_karaoke else "&H80000000"
    outline_col = cb if is_karaoke else "&H00000000"
    pos_y = int(1920 * (max(12, min(88, y_pct)) / 100.0))

    hdr = (
        "[Script Info]\nScriptType: v4.00+\nPlayResX: 1080\nPlayResY: 1920\n"
        "[V4+ Styles]\nFormat: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, OutlineColour, BackColour, "
        "Bold, Italic, Underline, StrikeOut, ScaleX, ScaleY, Spacing, Angle, BorderStyle, Outline, Shadow, Alignment, MarginL, MarginR, MarginV, Encoding\n"
        f"Style: Default,DejaVu Sans,{fsize},{cp},{ch},{outline_col},{back},-1,0,0,0,100,100,0,0,{bs},{osize},{sh},5,60,60,0,1\n"
        "[Events]\nFormat: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, Text\n"
    )
    cw = [w for w in words if w["end"] > st_c and w["start"] < en_c] or words[:12]
    evs = []
    pos_tag = f"{{\\pos(540,{pos_y})}}"
    for chunk in [cw[i:i+3] for i in range(0, len(cw), 3)]:
        if not chunk:
            continue
        if is_karaoke:
            k_parts = []
            for w in chunk:
                wt = re.sub(r"[^A-Za-z0-9'!?., -]", "", w["word"]).strip().upper()
                if not wt:
                    continue
                dur_cs = max(5, int(round((min(w["end"], en_c) - max(w["start"], st_c)) * 100)))
                k_parts.append(f"{{\\kf{dur_cs}\\c{ch}}}{wt}{{\\c{cp}}}")
            if k_parts:
                evs.append(f"Dialogue: 0,{fmt_t(chunk[0]['start']-st_c)},{fmt_t(chunk[-1]['end']-st_c)},Default,,0,0,0,,{pos_tag}{''.join(k_parts)}")
        elif mode == "Hormozi Pop (Word Highlight)":
            for i, aw in enumerate(chunk):
                parts = []
                for j, w in enumerate(chunk):
                    wt = re.sub(r"[^A-Za-z0-9'!?., -]", "", w["word"]).strip().upper()
                    if not wt:
                        continue
                    if j == i:
                        parts.append(f"{{\\c{ch}\\fscx114\\fscy114}}{wt}{{\\c{cp}\\fscx100\\fscy100}}")
                    else:
                        parts.append(wt)
                if parts:
                    evs.append(f"Dialogue: 0,{fmt_t(aw['start']-st_c)},{fmt_t(aw['end']-st_c)},Default,,0,0,0,,{pos_tag}{' '.join(parts)}")
        else:
            txt = " ".join(re.sub(r"[^A-Za-z0-9'!?., -]", "", w["word"]).strip().upper() for w in chunk).strip()
            if txt:
                evs.append(f"Dialogue: 0,{fmt_t(chunk[0]['start']-st_c)},{fmt_t(chunk[-1]['end']-st_c)},Default,,0,0,0,,{pos_tag}{txt}")
    open(path, "w", encoding="utf-8").write(hdr + "\n".join(evs))

def upd(jid, key, state, detail="", prog=None, eta=None):
    j = JOBS.get(jid)
    if not j:
        return
    for s in j["steps"]:
        if s["key"] == key:
            s["state"] = state
            if detail:
                s["detail"] = detail
    if prog is not None:
        j["progress"] = prog
    if eta is not None:
        j["eta"] = eta

def worker(jid, sid, key, count, min_len, max_len, layout, tx, ty, tw, th, bx, by, bw, bh, cstyle, cfont, csize, cy_pct, ctc, chc, cbc, pstyle):
    sdir = f"sessions/{sid}"
    ivid, iaud = f"{sdir}/input.mp4", f"{sdir}/audio.wav"
    step = "audio"
    try:
        upd(jid, "audio", "active", "Extracting 16kHz mono audio...", 10, 18)
        r = subprocess.run(["ffmpeg", "-y", "-i", ivid, "-vn", "-acodec", "pcm_s16le", "-ar", "16000", "-ac", "1", iaud], capture_output=True, text=True)
        if r.returncode != 0 or not os.path.exists(iaud):
            raise RuntimeError(r.stderr[-300:])
        upd(jid, "audio", "done", "Audio extracted", 25, 14)

        step = "whisper"
        upd(jid, "whisper", "active", "Transcribing speech...", 35, 10)
        with wave.open(iaud, "rb") as wf:
            anp = np.frombuffer(wf.readframes(wf.getnframes()), dtype=np.int16).astype(np.float32) / 32768.0
        segs, _ = whisper_model.transcribe(anp, word_timestamps=True)
        words, tlines = [], []
        for s in segs:
            tlines.append(f"[{s.start:.1f}s - {s.end:.1f}s] {s.text.strip()}")
            if s.words:
                for w in s.words:
                    words.append({"word": w.word, "start": w.start, "end": w.end})
        if not words:
            raise RuntimeError("No speech detected in audio.")
        upd(jid, "whisper", "done", f"Transcribed {len(words)} words", 55, 8)

        step = "gemini"
        upd(jid, "gemini", "active", "Finding viral hooks with Gemini...", 60, 6)
        client = genai.Client(api_key=key.strip())
        prompt = (
            f"Analyze this transcript and extract the top {count} complete, self-contained viral moments.\n"
            f"Each clip MUST have a natural start and end between {min_len}s and {max_len}s long (never cut mid-sentence).\n"
            f"Focus: {pstyle}.\n"
            f'Return ONLY a valid JSON array: [{{"title":"Hook Title","start":12.5,"end":46.0,"virality_score":94,"reason":"Why it hooks viewers"}}]\n'
            f"Transcript:\n" + "\n".join(tlines[:180])
        )
        resp, last_err = None, None
        for m_name in ["gemini-3.8-flash", "gemini-3.6-flash", "gemini-3.5-flash-lite", "gemini-3.1-flash-lite"]:
            try:
                resp = client.models.generate_content(model=m_name, contents=prompt)
                break
            except Exception as me:
                last_err = me
        if resp is None:
            for m in client.models.list():
                if "flash" in m.name and "generateContent" in (m.supported_actions or []):
                    resp = client.models.generate_content(model=m.name, contents=prompt)
                    break
        if resp is None:
            raise last_err
        meta = json.loads(re.sub(r"^```json|```$", "", resp.text.strip(), flags=re.MULTILINE).strip())[:count]
        upd(jid, "gemini", "done", f"Selected {len(meta)} viral moments", 68, len(meta) * 4)

        step = "render"
        ensure_font(cfont)
        tot = len(meta)
        for i, cm in enumerate(meta):
            st, en = float(cm["start"]), float(cm["end"])
            dur = en - st
            if dur < min_len:
                en = st + min_len
            elif dur > max_len:
                en = st + max_len
            actual_dur = round(en - st, 1)
            upd(jid, "render", "active", f"Rendering Clip {i+1}/{tot} ({actual_dur}s)...", 68 + int((i / tot) * 30), (tot - i) * 4)
            ass_p = os.path.abspath(f"{sdir}/clip_{i+1}.ass")
            fdir = os.path.abspath("fonts")
            out_p = f"{sdir}/clip_{i+1}.mp4"
            build_ass(words, st, en, cstyle, cfont, csize, cy_pct, ctc, chc, cbc, ass_p)
            sub_f = f"ass={ass_p}"
            if layout == "split":
                vf = (
                    f"[0:v]crop=iw*{tw}:ih*{th}:iw*{tx}:ih*{ty},scale=1080:960:force_original_aspect_ratio=increase,crop=1080:960[top];"
                    f"[0:v]crop=iw*{bw}:ih*{bh}:iw*{bx}:ih*{by},scale=1080:960:force_original_aspect_ratio=increase,crop=1080:960[bot];"
                    f"[top][bot]vstack=inputs=2,{sub_f}[v]"
                )
            elif layout == "dual":
                vf = (
                    f"[0:v]crop=iw*{tw}:ih*{th}:iw*{tx}:ih*{ty},scale=540:1920:force_original_aspect_ratio=increase,crop=540:1920[lft];"
                    f"[0:v]crop=iw*{bw}:ih*{bh}:iw*{bx}:ih*{by},scale=540:1920:force_original_aspect_ratio=increase,crop=540:1920[rgt];"
                    f"[lft][rgt]hstack=inputs=2,{sub_f}[v]"
                )
            elif layout == "blur":
                vf = (
                    f"[0:v]split=2[bg][fg];"
                    f"[bg]scale=1080:1920:force_original_aspect_ratio=increase,crop=1080:1920,boxblur=25:5[bgb];"
                    f"[fg]crop=iw*{tw}:ih*{th}:iw*{tx}:ih*{ty},scale=1080:-2[fgc];"
                    f"[bgb][fgc]overlay=(W-w)/2:(H-h)/2,{sub_f}[v]"
                )
            else:
                vf = f"[0:v]crop=iw*{tw}:ih*{th}:iw*{tx}:ih*{ty},scale=1080:1920:force_original_aspect_ratio=increase,crop=1080:1920,{sub_f}[v]"
            rv = subprocess.run(["ffmpeg", "-y", "-ss", str(st), "-to", str(en), "-i", ivid, "-filter_complex", vf, "-map", "[v]", "-map", "0:a?", "-c:v", "libx264", "-preset", "veryfast", "-crf", "22", "-c:a", "aac", out_p], capture_output=True, text=True)
            if rv.returncode != 0 or not os.path.exists(out_p):
                raise RuntimeError(rv.stderr[-300:])
            JOBS[jid]["clips"].append({"index": i + 1, "duration": actual_dur, "title": cm.get("title", f"Viral Clip #{i+1}"), "score": cm.get("virality_score", 92), "reason": cm.get("reason", "High retention hook"), "url": f"/sessions/{sid}/clip_{i+1}.mp4"})
            upd(jid, "render", "active", f"Rendered {i+1}/{tot} clips...", 68 + int(((i + 1) / tot) * 31), (tot - i - 1) * 4)

        upd(jid, "render", "done", f"All {tot}/{tot} clips ready!", 100, 0)
        JOBS[jid]["status"] = "done"
    except Exception as e:
        traceback.print_exc()
        upd(jid, step, "error", str(e))
        JOBS[jid]["status"] = "error"
        JOBS[jid]["error"] = f"[{step.upper()} STEP FAILED] {str(e)}"

@app.get("/", response_class=HTMLResponse)
async def home():
    return open("static/index.html", "r", encoding="utf-8").read()

@app.post("/api/prepare")
async def prepare(youtube_url: str = Form(""), cookies_text: str = Form(""), video_file: UploadFile = File(None)):
    sid = uuid.uuid4().hex[:10]
    sdir = f"sessions/{sid}"
    os.makedirs(sdir, exist_ok=True)
    ivid = f"{sdir}/input.mp4"
    try:
        if video_file and video_file.filename:
            open(ivid, "wb").write(await video_file.read())
        elif youtube_url.strip():
            if cookies_text.strip():
                with open("cookies.txt", "w", encoding="utf-8") as cf:
                    cf.write(cookies_text.strip() + "\n")
            subprocess.run(["yt-dlp", "-U"], capture_output=True, text=True, timeout=30)
            attempts = [
                ["yt-dlp", "--js-runtimes", "deno:/root/.deno/bin/deno", "--no-check-certificates", "--extractor-args", "youtube:player_client=android,ios,web", "-f", "best[ext=mp4][height<=1080]/best[ext=mp4]/best", "--merge-output-format", "mp4", "-o", ivid, youtube_url.strip()],
                ["yt-dlp", "--js-runtimes", "deno:/root/.deno/bin/deno", "--no-check-certificates", "-f", "best[ext=mp4]/best", "--merge-output-format", "mp4", "-o", ivid, youtube_url.strip()],
            ]
            last_err, ok = "", False
            for base_cmd in attempts:
                cmd = list(base_cmd)
                if os.path.exists("cookies.txt") and os.path.getsize("cookies.txt") > 10:
                    idx = cmd.index("--no-check-certificates") + 1
                    cmd[idx:idx] = ["--cookies", "cookies.txt"]
                r = subprocess.run(cmd, capture_output=True, text=True, timeout=120)
                if r.returncode == 0 and os.path.exists(ivid) and os.path.getsize(ivid) > 1000:
                    ok = True
                    break
                last_err = (r.stderr or r.stdout or "Download failed")[-240:]
            if not ok:
                return JSONResponse(status_code=400, content={"error": f"YouTube download failed ({last_err}). Try cookies.txt or Upload MP4."})
        else:
            return JSONResponse(status_code=400, content={"error": "Enter a YouTube URL or upload an MP4 file."})
        return {"session_id": sid, "video_url": f"/sessions/{sid}/input.mp4"}
    except Exception as e:
        return JSONResponse(status_code=500, content={"error": str(e)})

@app.post("/api/generate")
async def generate(
    session_id: str = Form(""), api_key: str = Form(""), clip_count: int = Form(3),
    min_len: int = Form(20), max_len: int = Form(45),
    layout: str = Form("single"), top_x: float = Form(0.25), top_y: float = Form(0.0), top_w: float = Form(0.5), top_h: float = Form(1.0),
    bot_x: float = Form(0.35), bot_y: float = Form(0.55), bot_w: float = Form(0.3), bot_h: float = Form(0.45),
    cap_style: str = Form("Hormozi Pop (Word Highlight)"), cap_font: str = Form("Montserrat"),
    cap_size: int = Form(76), cap_y_pct: int = Form(75),
    cap_text_col: str = Form("#FFFFFF"), cap_hi_col: str = Form("#FFE600"), cap_bg_col: str = Form("#000000"),
    prompt_style: str = Form("Viral Hooks & High Retention")
):
    if not session_id.strip():
        return JSONResponse(status_code=400, content={"error": "Please load your video in Step 1 first."})
    if not api_key.strip():
        return JSONResponse(status_code=400, content={"error": "Please enter your free Gemini API Key."})
    jid = uuid.uuid4().hex[:10]
    JOBS[jid] = {
        "status": "running", "progress": 5, "eta": 15 + clip_count * 4, "error": "", "clips": [],
        "steps": [
            {"key": "audio", "label": "1. Extract Audio Track", "state": "pending", "detail": "Waiting..."},
            {"key": "whisper", "label": "2. Transcribe Speech (Whisper AI)", "state": "pending", "detail": "Waiting..."},
            {"key": "gemini", "label": "3. Find Viral Hooks (Gemini AI)", "state": "pending", "detail": "Waiting..."},
            {"key": "render", "label": f"4. Render & Caption {clip_count} Clips", "state": "pending", "detail": "Waiting..."}
        ]
    }
    threading.Thread(
        target=worker,
        args=(
            jid, session_id, api_key, clip_count, min_len, max_len, layout,
            top_x, top_y, top_w, top_h, bot_x, bot_y, bot_w, bot_h,
            cap_style, cap_font, cap_size, cap_y_pct, cap_text_col, cap_hi_col, cap_bg_col, prompt_style
        ),
        daemon=True
    ).start()
    return {"job_id": jid}

@app.get("/api/status/{jid}")
async def status(jid: str):
    return JOBS.get(jid) or JSONResponse(status_code=404, content={"error": "Job not found"})
````
